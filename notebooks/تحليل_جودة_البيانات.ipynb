```python

import pandas as pd
import os

# رابط ملف البيانات على GitHub
url = "https://raw.githubusercontent.com/alwaylee/Data-quality-project/refs/heads/main/data/employees.csv"

# قراءة البيانات
df = pd.read_csv(url)

# حساب مؤشرات جودة البيانات
total_records = len(df)
duplicate_records = df.duplicated().sum()
missing_values = df.isnull().sum().sum()

# إنشاء التقرير
report = f"""
تقرير جودة البيانات
--------------------
إجمالي السجلات: {total_records}
عدد السجلات المكررة: {duplicate_records}
عدد القيم المفقودة: {missing_values}
"""

# إنشاء مجلد التقارير
os.makedirs("reports", exist_ok=True)

# حفظ التقرير
with open("reports/quality_report.txt", "w", encoding="utf-8") as file:
    file.write(report)

print(report)
print("تم إنشاء تقرير جودة البيانات بنجاح")
```

```python
# عرض عدد القيم المفقودة في كل عمود
missing_by_column = df.isnull().sum()

print("القيم المفقودة حسب الأعمدة:")
print(missing_by_column[missing_by_column > 0])

# عرض السجلات التي تحتوي على قيم مفقودة
print("\nالسجلات التي تحتوي على قيم مفقودة:")
display(df[df.isnull().any(axis=1)])
```

```python
# عرض أسماء الأعمدة وأنواع البيانات
print("أسماء الأعمدة:")
print(df.columns.tolist())

print("\nأنواع البيانات:")
print(df.dtypes)

print("\nأول خمسة سجلات:")
display(df.head())
```

```python

# حساب متوسط العمر والراتب
age_mean = df['Age'].mean()
salary_mean = df['Salary'].mean()

# تعويض القيم المفقودة بالمتوسط
df['Age'] = df['Age'].fillna(age_mean)
df['Salary'] = df['Salary'].fillna(salary_mean)

# عرض المتوسطات
print("متوسط العمر:", age_mean)
print("متوسط الراتب:", salary_mean)

# التحقق من نجاح المعالجة
print("\nالقيم المفقودة بعد المعالجة:")
print(df[['Age', 'Salary']].isnull().sum())

# عرض البيانات بعد المعالجة
display(df)

```

```python

import os

# إنشاء مجلد لحفظ البيانات النظيفة
os.makedirs("processed_data", exist_ok=True)

# حفظ البيانات بعد المعالجة
df.to_csv(
    "processed_data/employees_cleaned.csv",
    index=False
)

print("تم حفظ البيانات النظيفة بنجاح")

# التحقق من القيم المفقودة
print("إجمالي القيم المفقودة:", df.isnull().sum().sum())

# عرض أول خمسة سجلات
display(df.head())
```

```python

# تحديد الأعمدة المستخدمة للمقارنة
comparison_columns = [
    'Name',
    'Department',
    'Age',
    'Salary'
]

# اكتشاف السجلات المتشابهة
similar_records = df[
    df.duplicated(
        subset=comparison_columns,
        keep=False
    )
]

# عرض النتائج
print("عدد السجلات المتشابهة:")
print(len(similar_records))

print("\nالسجلات المتشابهة:")
display(similar_records)
```

```python

# تجميع السجلات المتشابهة
similar_groups = (
    similar_records
    .groupby(['Name', 'Department', 'Age', 'Salary'])
    .size()
    .reset_index(name='عدد السجلات')
)

print("مجموعات السجلات المتشابهة:")
display(similar_groups)

print("عدد مجموعات التشابه:", len(similar_groups))
print("إجمالي السجلات المتشابهة:", len(similar_records))
```

```python
# عرض جميع السجلات المتشابهة
duplicate_records = df[
    df.duplicated(
        subset=['Name', 'Department', 'Age', 'Salary'],
        keep=False
    )
]

display(duplicate_records)
```

```python
# الاحتفاظ بأول سجل وحذف التكرارات
df_cleaned = df.drop_duplicates(
    subset=['Name', 'Department', 'Age', 'Salary'],
    keep='first'
)

# عرض البيانات بعد المعالجة
print("عدد السجلات قبل المعالجة:", len(df))
print("عدد السجلات بعد المعالجة:", len(df_cleaned))
print("عدد السجلات التي تم حذفها:", len(df) - len(df_cleaned))

display(df_cleaned)
```

```python
# قياس جودة البيانات بعد المعالجة

print("عدد القيم المفقودة:")
print(df_cleaned.isnull().sum())

print("\nعدد السجلات المكررة:")
print(df_cleaned.duplicated(
    subset=['Name', 'Department', 'Age', 'Salary']
).sum())

print("\nعدد السجلات النهائي:")
print(len(df_cleaned))
```

```python
# فحص القيم غير المنطقية

print("الأعمار الأقل من 18 أو الأكبر من 65:")
display(df_cleaned[
    (df_cleaned['Age'] < 18) |
    (df_cleaned['Age'] > 65)
])

print("الرواتب الأقل من أو تساوي صفر:")
display(df_cleaned[
    df_cleaned['Salary'] <= 0
])

print("الأقسام الموجودة:")
print(df_cleaned['Department'].unique())
```

```python
# توحيد أسماء الأقسام

department_mapping = {
    'Human Resources': 'HR'
}

df_cleaned['Department'] = (
    df_cleaned['Department'].replace(department_mapping)
)

print("الأقسام بعد التوحيد:")
print(df_cleaned['Department'].unique())
```

```python
# معالجة الأعمار غير المنطقية

df_cleaned.loc[
    (df_cleaned['Age'] < 18) |
    (df_cleaned['Age'] > 65),
    'Age'
] = None

print("تمت معالجة الأعمار غير المنطقية")
```

```python
# التقرير النهائي

print("عدد السجلات:", len(df_cleaned))
print("القيم المفقودة:", df_cleaned.isnull().sum().sum())
print("السجلات المكررة:", df_cleaned.duplicated(
    subset=['Name', 'Department', 'Age', 'Salary']
).sum())

# حفظ البيانات النظيفة
df_cleaned.to_csv(
    'employees_cleaned.csv',
    index=False,
    encoding='utf-8-sig'
)

print("تم حفظ ملف البيانات النظيفة بنجاح")
```
